```python
import numpy as np
import matplotlib.pyplot as plt
path = "../../inputs/"
```

```python
### Uncorrelate parameters for more efficient sampling

# Load in the covariance matrices and best-fit parameters from Alex
c_sn = np.array([1.34,-0.314,0.033])
c_sn_bao = np.array([1.307,-0.282,0.021])
c_sn_bao_qso = np.array([1.059,-0.2264])
c_sn_bao_cmb = np.array([0.1110,-0.02638])
c_all = np.array([0.1102, -0.0256])
cov_sn = np.loadtxt(path+"inv_fisher_matrix_sn")
cov_sn_bao = np.loadtxt(path+"inv_fisher_matrix_sn_bao")
cov_sn_bao_qso = np.loadtxt(path+"inv_fisher_matrix_sn_bao_qso.txt")
cov_sn_bao_cmb = np.loadtxt(path+"inv_fisher_matrix_sn_bao_cmb")
cov_all = np.loadtxt(path+"inv_fisher_matrix_all.txt")

# Perform Cholesky decomposition
L_sn = np.linalg.cholesky(cov_sn)
L_sn_bao = np.linalg.cholesky(cov_sn_bao)
L_sn_bao_qso = np.linalg.cholesky(cov_sn_bao_qso)
L_sn_bao_cmb = np.linalg.cholesky(cov_sn_bao_cmb)
L_all = np.linalg.cholesky(cov_all)

# Uncorrelate parameters
c_sn_uncorr = np.linalg.inv(L_sn) @ c_sn
c_sn_bao_uncorr = np.linalg.inv(L_sn_bao) @ c_sn_bao
c_sn_bao_qso_uncorr = np.linalg.inv(L_sn_bao_qso) @ c_sn_bao_qso
c_sn_bao_cmb_uncorr = np.linalg.inv(L_sn_bao_cmb) @ c_sn_bao_cmb
c_all_uncorr = np.linalg.inv(L_all) @ c_all

print("Uncorrelated coefficients:")
print(f"SN                   : {c_sn_uncorr}")
print(f"SN + BAO             : {c_sn_bao_uncorr}")
print(f"SN + BAO + QSO       : {c_sn_bao_qso_uncorr}")
print(f"SN + BAO + CMB       : {c_sn_bao_cmb_uncorr}")
print(f"SN + BA0 + CMB + QSO : {c_all_uncorr}")
```

```python
# Re-correlate parameters
# You can input the best-fit coefficients from your run her to obtain the correlated coefficients
c_sn_corr = L_sn @ c_sn_uncorr
c_sn_bao_corr = L_sn_bao @ c_sn_bao_uncorr
c_sn_bao_cmb_corr = L_sn_bao_cmb @ c_sn_bao_cmb_uncorr
c_all_corr = L_all @ c_all_uncorr

print("Correlated coefficients:")
print(f"SN                   : {c_sn_corr}")
print(f"SN + BAO             : {c_sn_bao_corr}")
print(f"SN + BAO + CMB       : {c_sn_bao_cmb_corr}")
print(f"SN + BAO + CMB + QSO : {c_all_corr}")
```
