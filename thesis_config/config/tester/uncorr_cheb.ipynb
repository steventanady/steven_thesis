```python
c_sn = np.array([1.34,-0.314,0.033])
c_sn_bao = np.array([1.307,-0.282,0.021])
c_sn_bao_qso = np.array([1.059,-0.2264])
c_sn_bao_cmb = np.array([0.1110,-0.02638])
c_all = np.array([0.1102, -0.0256])
cov_sn = np.loadtxt(path+"inv_fisher_matrix_sn")
cov_sn_bao = np.loadtxt(path+"inv_fisher_matrix_sn_bao")
cov_sn_bao_qso = np.loadtxt(path+"inv_fisher_matrix_sn_bao_qso.txt")
cov_sn_bao_cmb = np.loadtxt(path+"inv_fisher_matrix_sn_bao_cmb")
cov_all = np.loadtxt(path+"inv_fisher_matrix_all.txt")

L_sn = np.linalg.cholesky(cov_sn)
L_sn_bao = np.linalg.cholesky(cov_sn_bao)
L_sn_bao_qso = np.linalg.cholesky(cov_sn_bao_qso)
L_sn_bao_cmb = np.linalg.cholesky(cov_sn_bao_cmb)
L_all = np.linalg.cholesky(cov_all)

c_sn = np.linalg.inv(L_sn) @ c_sn
c_sn_bao = np.linalg.inv(L_sn_bao) @ c_sn_bao
c_sn_bao_qso = np.linalg.inv(L_sn_bao_qso) @ c_sn_bao_qso
c_sn_bao_cmb = np.linalg.inv(L_sn_bao_cmb) @ c_sn_bao_cmb
c_all = np.linalg.inv(L_all) @ c_all

print(f"SN                   : {c_sn}")
print(f"SN + BAO             : {c_sn_bao}")
print(f"SN + BAO + QSO       : {c_sn_bao_qso}")
print(f"SN + BAO + CMB       : {c_sn_bao_cmb}")
print(f"SN + BA0 + CMB + QSO : {c_all}")
```

```python
import numpy as np

# Load Fisher matrix
F = np.loadtxt("fisher_matrix.txt")

# Eigen decomposition
eigvals, eigvecs = np.linalg.eig(F)

# Inverse sqrt of eigenvalues
s = 1 / np.sqrt(eigvals)
print(f"correlated errors : {s}")

# Rotate back to parameter basis
dp = np.abs(eigvecs @ s)

print(f"uncorrelated errors : {dp}")  # Matches dp in C++
```
